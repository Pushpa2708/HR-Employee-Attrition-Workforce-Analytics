# **HR Employee Attrition & Workforce Analytics**

Employee attrition is an important workforce challenge. Understanding why employees leave can help organisations identify retention problems and improve workforce management.

In this analysis, we explore an HR dataset containing 1,240 employee records. We investigate employee demographics, departmental attrition, salary, job levels, satisfaction, promotions and other workforce factors.

The analysis uses Python to examine patterns, relationships and possible factors associated with employee exits.

**Objective:** To identify meaningful workforce patterns and translate the findings into useful HR insights.

## 7. Attrition Distribution

Analyse the number and percentage of employees who stayed and left.

```python
attrition_count = df['attrition'].value_counts()

display(attrition_count)

print("\nAttrition Percentage:")
display((df['attrition'].value_counts(normalize=True) * 100).round(2))

sns.countplot(data=df, x='attrition')
plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.show()
```

**Interpretation:**

Out of 1,240 employees, 1,063 (85.73%) stayed with the organisation, while 177 (14.27%) left. Most employees remained, but the 177 exits require further investigation to understand workforce retention patterns.

## 8. Department-wise Attrition Analysis

Compare employee exits across departments and calculate the attrition rate for each department.

```python
dept_attrition = pd.crosstab(
    df['department'],
    df['attrition']
)

dept_attrition['Attrition_Rate'] = (
    dept_attrition['Yes'] / dept_attrition.sum(axis=1) * 100
).round(2)

display(dept_attrition.sort_values(
    by='Attrition_Rate', ascending=False
))
```

**Interpretation:**

Sales has the highest attrition rate (23.72%), followed by Operations (18.93%). HR has the lowest attrition rate (5.83%). This suggests that Sales and Operations may need greater attention to employee retention.

## 9. Attrition by Exit Reason

Examine the most frequently reported reasons employees leave the organisation.

```python
exit_reasons = df[df['attrition'] == 'Yes']['exit_reason'].value_counts()

display(exit_reasons)

exit_reasons.plot(kind='bar', figsize=(9, 5))
plt.title('Reasons for Employee Exit')
plt.xlabel('Exit Reason')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Interpretation:**

Better compensation is the most frequently reported exit reason (55 employees), followed by limited growth (41) and work-life balance (33). These three reasons account for 72.88% of all employee exits.

## 10. Salary Distribution by Attrition

Compare salary distributions between employees who stayed and those who left.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='attrition', y='annual_salary_inr')

plt.title('Salary Distribution by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Annual Salary (INR)')
plt.show()
```

**Interpretation:**

Employees who left generally have a lower median salary than employees who stayed. However, both groups show considerable salary variation and some high-salary outliers. This suggests a possible association between salary and attrition, but does not establish causation.


## 11. Satisfaction Score by Attrition

Compare employee satisfaction levels between employees who stayed and those who left.

```python
sns.boxplot(data=df, x='attrition', y='satisfaction_score')
plt.title('Employee Satisfaction by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Satisfaction Score')
plt.show()
```

**Interpretation:**

Employees who left had a lower median satisfaction score (2.8) compared to employees who stayed (3.6). This indicates that lower employee satisfaction is associated with higher attrition in the dataset.

## 12. Employee Tenure by Attrition

Compare employee tenure between those who stayed and those who left.

```python
sns.boxplot(data=df, x='attrition', y='tenure_years')
plt.title('Employee Tenure by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Tenure (Years)')
plt.show()
```

**Interpretation:**

Employees who left had a slightly lower median tenure (approximately 5 years) compared to employees who stayed (approximately 5.6 years). The difference is relatively small, suggesting that tenure alone may not explain employee attrition.

## 13. Correlation Analysis

Examine relationships between numerical employee attributes using a correlation matrix.

```python
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Matrix of HR Variables')
plt.show()
```

**Interpretation:**

The correlation matrix shows very weak relationships among most numerical HR variables. The highest positive correlation is between tenure and performance score (0.08), which is still very weak. Overall, no strong linear relationships are observed among these numerical variables.

## 14. Correlation of HR Variables with Attrition

Convert attrition into a numerical variable and examine its correlation with employee attributes.

```python
df['attrition_encoded'] = df['attrition'].map({
    'No': 0,
    'Yes': 1
})

attrition_corr = df.select_dtypes(include=np.number).corr()['attrition_encoded'].drop(
    'attrition_encoded'
).sort_values()

display(attrition_corr)

plt.figure(figsize=(8, 5))
sns.barplot(x=attrition_corr.values, y=attrition_corr.index)
plt.title('Correlation of Employee Attributes with Attrition')
plt.xlabel('Correlation Coefficient')
plt.ylabel('HR Variables')
plt.show()
```

**Interpretation:**

Satisfaction score has the strongest correlation with attrition (-0.327), indicating a moderate negative relationship. Performance score shows a weak negative relationship (-0.113), while salary, tenure and age have very weak negative relationships. Promotions and training hours show almost no linear relationship with attrition.

Employee satisfaction appears to be an important factor associated with attrition in this dataset. However, correlation does not establish causation.

## 15. Attrition by Age Group

Analyse employee attrition across different age groups.

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[18, 25, 35, 45, 60],
    labels=['18-25', '26-35', '36-45', '46-60']
)

age_attrition = pd.crosstab(
    df['age_group'],
    df['attrition'],
    normalize='index'
) * 100

display(age_attrition.round(2))

age_attrition['Yes'].plot(kind='bar', figsize=(8, 5))
plt.title('Attrition Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Interpretation:**

Employees aged 18–25 have the highest attrition rate (16.05%), followed by the 26–35 age group (15.38%). The 46–60 age group has the lowest rate (13.15%). This suggests that younger employees experience comparatively higher attrition in this dataset.

## 16. Attrition by Job Level

Compare employee attrition rates across different job levels.

```python
job_attrition = pd.crosstab(
    df['job_level'],
    df['attrition'],
    normalize='index'
) * 100

display(job_attrition.round(2))

job_attrition['Yes'].sort_values().plot(
    kind='bar', figsize=(8, 5)
)

plt.title('Attrition Rate by Job Level')
plt.xlabel('Job Level')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=45)
plt.show()
```

**Interpretation:**

Managers have the highest attrition rate (18.88%), while Senior employees have the lowest (10.85%). Mid-level employees also show a relatively high attrition rate of 15.30%. HR may need to investigate retention challenges at the Manager and Mid levels.

## 17. Attrition by Employment Type

Compare attrition rates across different employment types.

```python
employment_attrition = pd.crosstab(
    df['employment_type'],
    df['attrition'],
    normalize='index'
) * 100

display(employment_attrition.round(2))

employment_attrition['Yes'].plot(
    kind='bar', figsize=(8, 5)
)

plt.title('Attrition Rate by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Interpretation:**

Contract employees have the highest attrition rate (16%), followed by full-time employees (14.29%) and part-time employees (12.83%). This suggests that contract employees may need greater attention to retention.

## 17. Performance Score by Attrition

Examine whether employee performance scores differ between employees who stayed and those who left.

```python
sns.boxplot(data=df, x='attrition', y='performance_score')

plt.title('Performance Score by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Performance Score')
plt.show()
```

**Interpretation:**

Employees who left had a slightly lower median performance score (3.2) compared to employees who stayed (3.5). However, the difference is relatively small, and both groups have overlapping score distributions. Performance score alone does not explain employee attrition.

## 18. Promotions by Attrition

Compare the number of promotions received by employees who stayed and those who left.

```python
sns.boxplot(data=df, x='attrition', y='promotions')

plt.title('Promotions by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Number of Promotions')
plt.show()
```

**Interpretation:**

Both employees who stayed and those who left have a median of zero promotions. Their promotion distributions are very similar, suggesting that the number of promotions does not show a clear difference between the two groups in this dataset.

## 21. Recommendation Impact Assessment

This section measures how many employee exits are associated with the main areas identified in the analysis. It also suggests KPIs that HR can track to evaluate future improvements.

The percentages represent historical exit reasons, not guaranteed reductions in attrition.

```python
# Calculate the percentage of exits associated with each reason

total_exits = (df['attrition'] == 'Yes').sum()

impact = df[df['attrition'] == 'Yes']['exit_reason'].value_counts()

impact_assessment = pd.DataFrame({
    'Employees': impact,
    'Percentage_of_Exits': (impact / total_exits * 100).round(2)
})

display(impact_assessment)
```

##Interpretation
- Better compensation: 55 employees (31.07%) — highest reported reason.
- Limited growth: 41 employees (23.16%).
- Work-life balance: 33 employees (18.64%).
- Manager issues: 22 employees (12.43%).
The top three reasons account for 72.88% of all 177 exits.

```python
recommendation_impact = pd.DataFrame({
    'HR Recommendation': [
        'Review compensation',
        'Improve career growth',
        'Support work-life balance',
        'Improve manager practices'
    ],
    'Employees': [55, 41, 33, 22],
    'Percentage of Exits': [31.07, 23.16, 18.64, 12.43],
    'Suggested KPI': [
        'Voluntary attrition rate by salary range',
        'Internal promotion rate',
        'Employee work-life balance score',
        'Manager satisfaction score'
    ]
})

display(recommendation_impact)
```

## 22. Scenario analysis
Let's estimate how overall attrition could change if HR successfully retained some employees who left for the four reasons in our table.

```python
total_employees = len(df)
total_exits = (df['attrition'] == 'Yes').sum()

targeted_exits = recommendation_impact['Employees'].sum()

scenarios = pd.DataFrame({
    'Retention Scenario': ['10%', '20%', '30%']
})

scenarios['Employees Retained'] = (
    targeted_exits * scenarios['Retention Scenario'].str.rstrip('%').astype(float) / 100
).round().astype(int)

scenarios['Estimated Remaining Exits'] = (
    total_exits - scenarios['Employees Retained']
)

scenarios['Estimated Attrition Rate (%)'] = (
    scenarios['Estimated Remaining Exits'] / total_employees * 100
).round(2)

display(scenarios)
```

### Interpretation

The scenario analysis estimates how employee attrition could change if HR successfully retained some employees associated with the four identified exit reasons.

- **10% retention:** Approximately 15 employees could be retained, reducing the estimated attrition rate from 14.27% to 13.06%.
- **20% retention:** Approximately 30 employees could be retained, reducing the estimated attrition rate to 11.85%.
- **30% retention:** Approximately 45 employees could be retained, reducing the estimated attrition rate to 10.65%.

**Business Insight:** Even a 10% improvement in retaining employees associated with these exit reasons could reduce the overall attrition rate.

**Note:** These are hypothetical scenarios based on historical exit reasons, not guaranteed outcomes. Actual results must be measured after HR implements the recommendations.
